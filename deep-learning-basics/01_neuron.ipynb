```python
# 뉴런 하나 계산
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 뉴런 하나: 위쪽은 +, 아래쪽은 - 가중치
w = np.array([1,1,-1,-1])

# 편향
b = 0


for name, a in  [("위 밝고 아래 어두움", [0.9, 0.9, 0.1, 0.1]),
                ("전부 밝음",           [0.9, 0.9, 0.9, 0.9]),
                ("위 어둡고 아래 밝음", [0.1, 0.1, 0.9, 0.9])]:
    z = np.dot(w,a) + b    # 가중합 + 편향
    print(name, round(sigmoid(z),3))
```

```python
# 뉴런 두개 계산
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 뉴런 두개
W = np.array([[2,2,-1,-1],[-1,-1,2,2]])

# 편향
b = np.array([0,0])




for a in  [[0.9, 0.9, 0.1, 0.1],
            [0.9, 0.9, 0.9, 0.9],
            [0.1, 0.1, 0.9, 0.9]]:
    print(sigmoid(W @ a + b))
```
